```python
# Skapa en lista av objekt som ska manipuleras likadant.
def normalize_points(points, scale_factor = 1.0, axis = 0):
    points *= scale_factor

    points[:,0] -= points[:,0].min()
    points[:,1] -= points[:,1].min()
    points[:,2] -= points[:,2].min()

    d = points[:,axis].max()
    points[:,0] /= d
    points[:,1] /= d
    points[:,2] /= d

normalize_points(obj_polydata1.points)
obj_polydata2 = obj_polydata1.copy()
obj_polydata2.points[:,0] += 1.1
group1_multiblock = pv.MultiBlock()
group1_multiblock.append(obj_polydata1)
group1_multiblock.append(obj_polydata2)
# group1_multiblock.plot()
```

```python
# skapa all_points som innehåller alla punkter i alla block i gruppen
total_points_group1 = 0
for block in group1_multiblock:
    total_points_group1 += block.points.shape[0]
all_points_group1 = np.zeros((total_points_group1, 3))

# Gör så att varje objekts points är views av rätt del av all_points
current_index = 0
for block in group1_multiblock:
    num_points = block.points.shape[0]
    all_points_group1[current_index:current_index+num_points] = block.points.copy()
    block.points = all_points_group1[current_index:current_index+num_points]
    current_index += num_points
```

```python
# Testa Object3D och Group3D
import pyvista_anim.object3D as object3D
star_polydata = pv.read('D:/Code/math_visualizations/3Dmodel/tinker.obj')
sphere_polydata = pv.Sphere(radius=0.5, center=(0,0,0), theta_resolution=16, phi_resolution=16)
dodeca_multiblock = pv.read("D:/Code/math_visualizations/3Dmodel/dodeca.glb")
dodeca_polydata = dodeca_multiblock[0][0][0][0]
object3D.normalize_points(star_polydata.points)
object3D.normalize_points(sphere_polydata.points)
object3D.normalize_points(dodeca_polydata.points)

star_object3D = object3D.Object3D(star_polydata)
sphere_object3D = object3D.Object3D(sphere_polydata)
dodeca_object3D = object3D.Object3D(dodeca_polydata)

sphere_object3D.points += np.array([1.5, 0, 0])
dodeca_object3D.points += np.array([0, 1.5, 0])

group3D = object3D.Group3D([sphere_object3D, star_object3D, dodeca_object3D])
group3D.generate_all_points()
group3D.all_points[:,1] *= 2
group3D.multiblock.plot()
```

```python
# Testa nu att manipulera all_points_group1 och se att det påverkar alla block i gruppen
all_points_group1[:,1] += all_points_group1[:,0] * 1.5
group1_multiblock.plot()
# Fungerar! Alla block i gruppen har påverkats av förändringen i all_points_group1.
```
